```python
# read in mock data
data = pd.read_csv("data/ot_sample_results.txt", sep=",", header=0)
data
```

```python
# summarise the data
n_unfair = data.unfair.sum()
n_samples = data.sample_size.sum()
unfair_rate = round(n_unfair / n_samples, 3)
fair_rate = 1 - unfair_rate
print(f"Fair outcome rate for the period: {fair_rate}")
```
